# Notebook 01: Formula 1 Telemetry Data Acquisition
### Machine Learning Laboratory Mini-Project
**Objective**: Demonstrate programmatic extraction and disk caching of high-frequency Formula 1 timing and telemetry data using the `FastF1` API.

---

```python
import os
import fastf1
import pandas as pd
import matplotlib.pyplot as plt

# Configure disk caching to minimize network overhead and ensure reproducibility
CACHE_DIR = '../backend/data/fastf1_cache'
os.makedirs(CACHE_DIR, exist_ok=True)
fastf1.Cache.enable_cache(CACHE_DIR)
print(f'FastF1 Cache enabled at: {CACHE_DIR}')
```

## 1. Load Session Data
We load the 2023 Italian Grand Prix (Monza) Race session.


```python
season = 2023
grand_prix = 'Monza'
session_type = 'R'

session = fastf1.get_session(season, grand_prix, session_type)
session.load(laps=True, telemetry=True, weather=False)

print(f'Session loaded: {session.event["EventName"]}')
print(f'Total session laps recorded: {len(session.laps)}')
```

## 2. Extract Driver Laps & Fastest Lap Telemetry
Extract lap timing information and inspect available telemetry channels (`Speed`, `Throttle`, `Brake`, `nGear`, `RPM`, `DRS`).


```python
# Inspect Verstappen and Hamilton laps
ver_laps = session.laps.pick_drivers('VER')
ver_fastest = ver_laps.pick_fastest()
telemetry = ver_fastest.get_telemetry().add_distance()

print(f'Fastest Lap Time (VER): {ver_fastest["LapTime"]}')
print('Telemetry Columns:', list(telemetry.columns))
telemetry[['Distance', 'Speed', 'Throttle', 'Brake', 'nGear', 'RPM']].head()
```
