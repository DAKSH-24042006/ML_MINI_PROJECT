# Notebook 03: Exploratory Data Analysis (EDA)
### Machine Learning Laboratory Mini-Project
**Objective**: Analyze telemetry dynamics: Speed vs Distance, Throttle profiles, Correlation heatmap between telemetry attributes and lap performance.

---

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from backend.app.services.fastf1_service import FastF1DataService
service = FastF1DataService()
df = service.load_session_data(season=2023, grand_prix='Monza')['features_df']
print(f'Dataset shape: {df.shape}')
```

## 1. High-Frequency Distance-Aligned Telemetry Trace
Compare Speed and Throttle traces between Max Verstappen (VER) and Lewis Hamilton (HAM) across Monza.

```python
trace = service.get_telemetry_trace_comparison(2023, 'Monza', 'R', 'VER', 'HAM')
tel_df = pd.DataFrame(trace['telemetry_points'])

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 6), sharex=True)
ax1.plot(tel_df['distance'], tel_df['VER_speed'], label='VER Speed', color='#e10600', lw=2)
ax1.plot(tel_df['distance'], tel_df['HAM_speed'], label='HAM Speed', color='#00d2be', lw=1.8)
ax1.set_ylabel('Speed (km/h)')
ax1.set_title('Monza Speed Trace Comparison')
ax1.legend()
ax1.grid(True, alpha=0.3)

ax2.plot(tel_df['distance'], tel_df['VER_throttle'], label='VER Throttle %', color='#e10600', lw=1.5)
ax2.plot(tel_df['distance'], tel_df['HAM_throttle'], label='HAM Throttle %', color='#00d2be', lw=1.5)
ax2.set_xlabel('Distance along Track (m)')
ax2.set_ylabel('Throttle %')
ax2.legend()
ax2.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## 2. Correlation Matrix Heatmap
Examine linear relationships between numerical features (`avg_speed`, `full_throttle_pct`, `brake_pct`, `tyre_age`) and target `lap_time`.

```python
numeric_cols = ['avg_speed', 'max_speed', 'avg_throttle', 'full_throttle_pct', 'brake_pct', 'braking_intensity_avg', 'tyre_age', 'lap_time']
corr = df[numeric_cols].corr()

plt.figure(figsize=(9, 7))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f', vmin=-1, vmax=1, linewidths=0.5)
plt.title('Correlation Heatmap: Telemetry Features vs Lap Time')
plt.show()
```

## 3. Key Findings
- Strong negative correlation ($r \approx -0.85$) between `avg_speed` and `lap_time` (higher speed = lower lap time).
- Positive correlation between `brake_pct` and `lap_time` (more time on brakes reduces average speed).
- `tyre_age` demonstrates progressive lap-time degradation.
